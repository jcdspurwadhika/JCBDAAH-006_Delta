# **BACKGROUND**

**PENDAHULUAN**

Bank adalah lembaga keuangan yang memberikan layanan jasa keuangan kepada masyarakat luas yang memiliki fungsi krusial dalam menggerakkan perekonomian masyarakat. Sama seperti badan usaha maupun bentuk bisnis lainnya, tujuan utama bank adalah untuk menghasilkan profit bagi dirinya sendiri. Kegiatan bisnis bank adalah menghimpun dana dari masyarakat (seperti deposito, atau penerbitan surat hutang) maupun dana dari luar (pembiayaan eksternal) kemudian berperan sebagai financial intermediary yang menjembatani kepada mereka yang membutuhkan bantuan finansial (dalam hal ini pemberian pinjaman) baik perorangan maupun institusi yang kemudian dikenakan interest rate, dan interest rate inilah yang menjadi salah satu sumber pendapatan bank itu sendiri.

Agar suatu bank bisa memberikan manfaat finansial kepada masyarakat luas, bank harus menghimpun dana dari masyarakat kemudian menyalurkannya kembali dalam bentuk pinjaman serta produk finansial lainnya kepada yang membutuhkan. Tujuan utama dari marketing adalah untuk itu, mencari nasabah yang potensial agar tertarik untuk mengambil produk bank dalam rangka menghimpun dana dari masyarakat, salah satu bentuk marketing yang dilakukan adalah direct marketing melalui telepon (selanjutnya disebut telemarketing) untuk menawarkan produk perbankan kepada calon nasabah.

Banco de Portugal adalah Bank Sentral Nasional Republik Portugal yang beroperasi sebagai member Eurosystem. Meskipun European Central Bank (ECB) telah mengambil alih fungsi penentu utama kebijakan moneter, namun Banco de Portugal tetap menjadi regulator perbankan dan otoritas keuangan utama di negara tersebut. Sepanjang tahun 2000 - 2007 Banco de Portugal mengalami pertumbuhan kredit (pinjaman) yang kuat didorong oleh suku bunga yang rendah, laju pertumbuhan kredit ini membuat pendanaan bank menjadi sangat tergantung pada pembiayaan eksternal (International interbank funding) hingga 32% dari total aset.

Ketergantungan Banco de Portugal terhadap pembiayaan eksternal ini ternyata membuat strategi sistem perbankan menjadi sangat rentan oleh gejolak eksternal. Kolom `age` hanya berisi nilai angka numerik dari umur nasabah saja. Perlu dilakukan **pengelompokan umur nasabah** berdasarkan rentang usia. Untuk itu dibuat kolom baru `age group` yang berisi kategori rentang usia.

Referensi untuk mengambil rentang/kelompok usia

https://sepakat-demo.bappenas.go.id/km-wiki/Kelompok_Usia

https://www.populationpyramids.org/portugal

```python
import numpy as np
import pandas as pd

# 1. Definisikan kondisi rentang usia
conditions = [
    df_bank['age'] < 25,  # <25
    (df_bank['age'] >= 25) & (df_bank['age'] <= 35),  # 25-35
    (df_bank['age'] >= 36) & (df_bank['age'] <= 55),  # 36-55
    (df_bank['age'] >= 56) & (df_bank['age'] <= 60),  # 56-60
    df_bank['age'] > 60,  # >60
]

# 2. Definisikan label angka rentang usia
choices = ['<25', '25-35', '36-55', '56-60', '>60']

# 3. Buat array nilai kelompok usia
age_group_series = np.select(conditions, choices, default='Unknown')

# 4. Hapus kolom 'age group' lama jika sudah pernah dibuat
if 'age group' in df_bank.columns:
  df_bank.drop(columns=['age group'], inplace=True)

# 5. Pindahkan kolom 'age group' tepat setelah kolom 'age'
age_idx = df_bank.columns.get_loc('age')
df_bank.insert(age_idx + 1, 'age group', age_group_series)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Pastikan kolom 'y' diubah menjadi angka 1 (yes) dan 0 (no)
if 'y' in df_bank.columns:
    df_bank['target'] = df_bank['y'].map({'yes': 1, 'no': 0})

# 2. Hitung total nasabah (count) dan total yang 'yes' (sum) berdasarkan pekerjaan
job_conv = df_bank.groupby('job').agg(
    total_nasabah=('target', 'count'),
    total_yes=('target', 'sum')
).reset_index()

# 3. Hitung persentase dari total SEMUA nasabah yang berkata "yes" (Proporsi keseluruhan)
total_seluruh_yes = job_conv['total_yes'].sum()
job_conv['persen_dari_total_yes'] = (job_conv['total_yes'] / total_seluruh_yes) * 100

# 4. Hitung Conversion Rate per pekerjaan
job_conv['conversion_rate'] = (job_conv['total_yes'] / job_conv['total_nasabah']) * 100

# 5. Urutkan berdasarkan angka absolut jumlah konversi tertinggi agar grafiknya rapi
job_conv = job_conv.sort_values('total_yes', ascending=False)

# 6. Membuat Visualisasi
sns.set_theme(style='whitegrid')
fig, ax = plt.subplots(figsize=(15, 8))

# Membuat grafik batang horizontal
sns.barplot(
    data=job_conv,
    x='total_yes',
    y='job',
    palette='viridis',
    ax=ax
)

ax.set_title('Jumlah Nasabah Konversi (Yes) Berdasarkan Pekerjaan', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Jumlah Nasabah yang Sukses (Angka Absolut)', fontsize=12, fontweight='bold')
ax.set_ylabel('Pekerjaan (Job)', fontsize=12, fontweight='bold')

# 7. Menambahkan teks gabungan (Angka + Persen 1 di sebelah kiri koma, Persen 2 di dalam kurung)
for i, row in enumerate(job_conv.itertuples()):
    # Format teks: " 1351 (29.12%)  (1351 sukses dari 10418, 12.97%)"
    teks_luar = f" {row.total_yes} ({row.persen_dari_total_yes:.2f}%)  ({row.total_yes} sukses dari {row.total_nasabah}, {row.conversion_rate:.2f}%)"
    ax.text(
        row.total_yes + 15, # Posisi X (sedikit di kanan ujung bar)
        i,                  # Posisi Y (sejajar dengan bar)
        teks_luar,
        va='center',
        ha='left',
        fontsize=11,
        color='black'
    )

# Memperlebar sumbu X agar teks yang cukup panjang tidak terpotong tepi gambar
ax.set_xlim(0, job_conv['total_yes'].max() * 1.5)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Urutan kategori status pernikahan dan kelompok sumbu X
marital_order = ['married', 'single', 'divorced', 'unknown']
group_order = ['Total', 'y = yes']

# Menghitung jumlah per kelompok
total_series = df_bank['marital'].value_counts()
yes_series = df_bank[df_bank['y'] == 'yes']['marital'].value_counts()

total_sum = total_series.sum()
yes_sum = yes_series.sum()

data = []

# Grup 1: Total (persentase dari total seluruh nasabah = 100%)
for cat in marital_order:
    count = total_series.get(cat, 0)
    pct = (count / total_sum) * 100 if total_sum > 0 else 0
    data.append({'Grup': 'Total', 'Marital': cat, 'Jumlah': count, 'Persentase': pct})

# Grup 2: y = yes (persentase dari total konversi yes = 100%)
for cat in marital_order:
    count = yes_series.get(cat, 0)
    pct = (count / yes_sum) * 100 if yes_sum > 0 else 0
    data.append({'Grup': 'y = yes', 'Marital': cat, 'Jumlah': count, 'Persentase': pct})

df_plot = pd.DataFrame(data)

# 4. Membuat Visualisasi
sns.set_theme(style='whitegrid')
fig, ax = plt.subplots(figsize=(12, 7))

# Warna masing-masing status pernikahan
palette = {
    'married': '#1f77b4',
    'single': '#2ca02c',
    'divorced': '#ff7f0e',
    'unknown': '#9467bd'
}

sns.barplot(
    data=df_plot,
    x='Grup',
    y='Jumlah',
    hue='Marital',
    hue_order=marital_order,
    palette=palette,
    ax=ax
)

ax.set_title('Distribusi Status Pernikahan (marital) pada Total dan y = yes', fontsize=15, fontweight='bold', pad=20)
ax.set_xlabel('Kelompok Data', fontsize=12, fontweight='bold')
ax.set_ylabel('Jumlah Nasabah (Angka Absolut)', fontsize=12, fontweight='bold')

# 5. Menambahkan Teks Anotasi Angka & Persen di atas bar
for i, container in enumerate(ax.containers):
    cat_hue = marital_order[i]
    for j, p in enumerate(container):
        grup_name = group_order[j]
        val = p.get_height()

        matched = df_plot[(df_plot['Marital'] == cat_hue) & (df_plot['Grup'] == grup_name)]
        pct_val = matched['Persentase'].values[0] if not matched.empty else 0
        teks = f"{int(val)}\n({pct_val:.2f}%)" if val > 0 else "0\n(0.00%)"

        ax.annotate(
            teks,
            (p.get_x() + p.get_width() / 2., max(val, 0)),
            ha='center', va='bottom',
            fontsize=10, color='black', fontweight='500',
            xytext=(0, 5),
            textcoords='offset points'
        )

# Memperluas batas sumbu Y dan mengatur legend
ax.set_ylim(0, df_plot['Jumlah'].max() * 1.15)
plt.legend(title='Status Pernikahan', loc='upper right')

plt.tight_layout()
plt.show()
```

Selanjutnya dilakukan analisa berdasarkan `age group` terhadap **conversion rate**

**Conversion rate `y` = yes berdasarkan `age group`**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Menyiapkan metrik konversi (Pastikan target biner sudah ada)
if 'y' in df_bank.columns and 'target' not in df_bank.columns:
    df_bank['target'] = df_bank['y'].map({'yes': 1, 'no': 0})

# 2. Hitung total nasabah dan total yang 'yes' berdasarkan 'age group' di dataset Anda
age_conv = df_bank.groupby('age group', observed=True).agg(
    total_nasabah=('target', 'count'),
    total_yes=('target', 'sum')
).reset_index()

# 3. Hitung dua jenis persentase yang diminta
total_seluruh_yes = age_conv['total_yes'].sum()
age_conv['persen_dari_total_yes'] = (age_conv['total_yes'] / total_seluruh_yes) * 100
age_conv['conversion_rate'] = (age_conv['total_yes'] / age_conv['total_nasabah']) * 100

# 4. KUNCI PERBAIKAN: Definisikan urutan secara eksplisit
# Tuliskan persis seperti nama kategori di dataset Anda (Tertua ke Termuda)
order_kategori = ['>60', '56-60', '36-55', '25-35', '<25']

# 5. Membuat Visualisasi
sns.set_theme(style='whitegrid')
fig, ax = plt.subplots(figsize=(15, 8))

# Masukkan 'order_kategori' ke dalam parameter 'order'
sns.barplot(
    data=age_conv,
    x='total_yes',
    y='age group',
    palette='viridis',
    ax=ax,
    order=order_kategori
)

ax.set_title('Jumlah Nasabah Berdasarkan Kelompok Usia', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Jumlah Nasabah yang setuju', fontsize=12, fontweight='bold')
ax.set_ylabel('Kelompok Usia (Age Group)', fontsize=12, fontweight='bold')

# 6. Menambahkan teks anotasi
for i, grup in enumerate(order_kategori):
    # Cari baris spesifik untuk grup tersebut
    row = age_conv[age_conv['age group'] == grup]

    # Jika datanya ada, tampilkan teks
    if not row.empty:
        row = row.iloc[0]
        teks_luar = f" {row.total_yes} ({row.persen_dari_total_yes:.2f}%)  ({row.total_yes} sukses dari {row.total_nasabah}, {row.conversion_rate:.2f}%)"

        ax.text(
            row.total_yes + 15,
            i,
            teks_luar,
            va='center',
            ha='left',
            fontsize=11,
            color='black'
        )

ax.set_xlim(0, age_conv['total_yes'].max() * 1.5)

plt.tight_layout()
plt.show()
```

Keterangan

Terdapat 3 label angka untuk tiap chart
- angka menunjukkan kuantitas yang dikonversi
- angka persen pertama menunjukkan persen komposisi dari total yang dikonversi
- angka persen kedua menunjukkan persen komposisi dari total nasabah



Temuan

- Kelompok high volume, low conversion: rentang `25-35` dan `36-55` didominasi oleh usia produktif yang menyumbang jumlah secara kuantitas paling besar. Namun, tingkat konversi (conversion rate) di dalam kelompok mereka sendiri adalah 8-12%.

- Kelompok low volume, high conversion: rentang `<25` dan `>60` memiliki conversion rate yang tinggi, terutama usia >60 mencapai 45,54%. Namun secara kuantitas mereka sedikit jumlahnya.

- Sedangkan kelompok `56-60` tidak termasuk ke dalam keduanya, karena tergolong low volume, low conversion. Tapi jika dilihat dari conversion rate nya ~12% tidak berbeda jauh dengan kelompok pertama, kita bisa memasukkan kelompok ini pada kelompok pertama.

**Performa Telemarketing dan conversion rate sepanjang 2008-2010**

Apabila melihat dari diagram sebelumnya yang memperlihatkan **total conversion hanya mencapai 11,3%.** Sekarang kita analisa lebih mendetail dengan cara mem-*breakdown* dari awal hingga akhir secara kronologis.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load data
# df_bank = pd.read_csv('bank_cleaned.csv', sep=';')

month_map = {'mar': 3, 'apr': 4, 'may': 5, 'jun': 6, 'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12}
month_names_id = {3: 'Mar', 4: 'Apr', 5: 'Mei', 6: 'Jun', 7: 'Jul', 8: 'Agu', 9: 'Sep', 10: 'Okt', 11: 'Nov', 12: 'Des'}

df_bank['month_num'] = df_bank['month'].map(month_map)

monthly = df_bank.groupby(['year', 'month_num']).agg(
    total=('y', 'count'),
    yes=('y', lambda x: (x == 'yes').sum())
).reset_index()

monthly['conversion_rate'] = (monthly['yes'] / monthly['total']) * 100
monthly = monthly.sort_values(['year', 'month_num']).reset_index(drop=True)
monthly['period'] = monthly['month_num'].map(month_names_id) + ' ' + monthly['year'].astype(str)

# Chart creation
fig, ax1 = plt.subplots(figsize=(18, 9))

x = np.arange(len(monthly))
width = 0.38

# Primary Y-axis: Bars
rects1 = ax1.bar(x - width/2, monthly['total'], width, label='Total Telemarketing', color='#2b5c8f', alpha=0.9)
rects2 = ax1.bar(x + width/2, monthly['yes'], width, label='Berhasil (y = yes)', color='#27ae60', alpha=0.9)

ax1.set_ylabel('Jumlah Kontak / Panggilan', fontsize=12, weight='bold', color='#2b5c8f')
ax1.set_xlabel('Periode', fontsize=12, weight='bold', labelpad=10)
ax1.set_xticks(x)
ax1.set_xticklabels(monthly['period'], rotation=45, ha='right', fontsize=10)
ax1.grid(True, linestyle='--', alpha=0.3, axis='y')

# Secondary Y-axis: Conversion Rate Line
ax2 = ax1.twinx()
line = ax2.plot(x, monthly['conversion_rate'], color='#e74c3c', marker='o', linewidth=2.5, markersize=6, label='Conversion Rate (%)')
ax2.set_ylabel('Conversion Rate (%)', fontsize=12, weight='bold', color='#e74c3c')
ax2.grid(False)
ax2.set_ylim(0, 115) # Expand a bit to accommodate 2-line annotation boxes

# Set y limit for ax1
ax1.set_ylim(0, max(monthly['total']) * 1.25)

# Annotations
for i in range(len(monthly)):
    tot = monthly['total'].iloc[i]
    y_val = monthly['yes'].iloc[i]
    rate = monthly['conversion_rate'].iloc[i]

    # Total count label
    ax1.annotate(f"{tot:,}".replace(',', '.'),
                 (x[i] - width/2, tot),
                 ha='center', va='bottom', fontsize=8, rotation=90, xytext=(0, 3), textcoords='offset points')

    # Success count label
    ax1.annotate(f"{y_val:,}".replace(',', '.'),
                 (x[i] + width/2, y_val),
                 ha='center', va='bottom', fontsize=8, rotation=90, xytext=(0, 3), textcoords='offset points')

    # Conversion rate label on line point with success count ABOVE percentage
    success_str = f"{y_val:,}".replace(',', '.')
    label_text = f"{success_str}\n({rate:.1f}%)"

    ax2.annotate(label_text,
                 (x[i], rate),
                 ha='center', va='bottom', fontsize=8, weight='bold', color='#c0392b',
                 xytext=(0, 6), textcoords='offset points',
                 bbox=dict(boxstyle='round,pad=0.25', facecolor='white', edgecolor='#e74c3c', alpha=0.9, linewidth=0.6))

# Combine legends
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left', fontsize=11, frameon=True, facecolor='white', edgecolor='none', framealpha=0.9)

plt.title('Performa telemarketing sepanjang Mei 2008 - Nov 2010', fontsize=15, weight='bold', pad=20)
plt.tight_layout()
# plt.savefig('updated_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

Dari diagram di atas terlihat bahwa data menjadi sangat tidak seimbang, terlihat banyak sekali bulan-bulan yang menghasilkan conversion rate >11,3% bahkan mencapai di atas 50%.

Rendahnya conversion rate pada rentang bulan Mei–Agustus 2008, November 2008, serta April–Mei 2009 diperkirakan Bank merasa panik pada awal krisis global terjadi dan langsung menelepon secara besar-besaran dalam rangka mencari sumber pendanaan dari deposito.

- Pada periode rendah tersebut, bank mengeksekusi total 35.851 panggilan (sekitar ~87% dari seluruh data), namun hanya menghasilkan rata-rata konversi 6,3%.

- Sebagai perbandingan, pada sisa bulan lainnya (seperti Mar, Sep, Okt, Des), terlihat Bank mengubah taktik. Mereka hanya melakukan total 5.321 panggilan, namun angka keberhasilannya melonjak drastis hingga rata-rata 44,7%.

Namun besarnya kuantitas total panggilan (35.851 panggilan) yang menghasilkan conversion rate membuat performa keberhasilan di bulan-bulan lainnya menjadi terkamuflase akibat data yang tidak seimbang (imbalance) tersebut.

Selanjutnya akan dilakukan analisis lebih lanjut faktor apa yang menyebabkan peningkatan conversion rate

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# Pemetaan Bulan untuk Urutan Kronologis
month_map = {
    'mar': 3,
    'apr': 4,
    'may': 5,
    'jun': 6,
    'jul': 7,
    'aug': 8,
    'sep': 9,
    'oct': 10,
    'nov': 11,
    'dec': 12,
}
month_names_id = {
    3: 'Mar',
    4: 'Apr',
    5: 'Mei',
    6: 'Jun',
    7: 'Jul',
    8: 'Agu',
    9: 'Sep',
    10: 'Okt',
    11: 'Nov',
    12: 'Des',
}

df_bank['month_num'] = df_bank['month'].map(month_map)

# 2. Agregasi per Tahun & Bulan berdasarkan previous (0 vs >0)
monthly = (
    df_bank.groupby(['year', 'month_num'])
    .agg(
        total_prev0=('previous', lambda x: (x == 0).sum()),
        total_prev_gt0=('previous', lambda x: (x > 0).sum()),
        yes_prev0=(
            'y',
            lambda x: (
                (df_bank.loc[x.index, 'previous'] == 0) & (x == 'yes')
            ).sum(),
        ),
        yes_prev_gt0=(
            'y',
            lambda x: (
                (df_bank.loc[x.index, 'previous'] > 0) & (x == 'yes')
            ).sum(),
        ),
        total=('y', 'count'),
        yes_total=('y', lambda x: (x == 'yes').sum()),
    )
    .reset_index()
)

monthly['conversion_rate'] = (monthly['yes_total'] / monthly['total']) * 100
monthly = monthly.sort_values(['year', 'month_num']).reset_index(drop=True)
monthly['period'] = (
    monthly['month_num'].map(month_names_id)
    + ' '
    + monthly['year'].astype(str)
)

# 3. Chart Creation
fig, ax1 = plt.subplots(figsize=(18, 9))

x = np.arange(len(monthly))
width = 0.38

# Primary Y-axis: Batang previous = 0 vs previous > 0
rects1 = ax1.bar(
    x - width / 2,
    monthly['total_prev0'],
    width,
    label='previous = 0 (Cold Leads Murni)',
    color='#1f77b4',
    alpha=0.9,
)
rects2 = ax1.bar(
    x + width / 2,
    monthly['total_prev_gt0'],
    width,
    label='previous > 0 (Pernah Dihubungi Sebelumnya)',
    color='#ff7f0e',
    alpha=0.9,
)

ax1.set_ylabel(
    'Jumlah Kontak / Panggilan (previous)',
    fontsize=12,
    weight='bold',
    color='#2c3e50',
)
ax1.set_xlabel('Periode', fontsize=12, weight='bold', labelpad=10)
ax1.set_xticks(x)
ax1.set_xticklabels(monthly['period'], rotation=45, ha='right', fontsize=10)
ax1.grid(True, linestyle='--', alpha=0.3, axis='y')

# Secondary Y-axis: Conversion Rate Line (%)
ax2 = ax1.twinx()
line = ax2.plot(
    x,
    monthly['conversion_rate'],
    color='#e74c3c',
    marker='o',
    linewidth=2.5,
    markersize=6,
    label='Conversion Rate (%)',
)
ax2.set_ylabel(
    'Conversion Rate (%)', fontsize=12, weight='bold', color='#e74c3c'
)
ax2.grid(False)
ax2.set_ylim(0, 115)

# Set y limit for ax1
ax1.set_ylim(0, max(monthly['total_prev0']) * 1.25)

# 4. Annotations
for i in range(len(monthly)):
  t_p0 = monthly['total_prev0'].iloc[i]
  y_p0 = monthly['yes_prev0'].iloc[i]
  t_pg0 = monthly['total_prev_gt0'].iloc[i]
  y_pg0 = monthly['yes_prev_gt0'].iloc[i]
  rate = monthly['conversion_rate'].iloc[i]

  # Label Batang Biru (previous = 0): Format "Total (Berhasil)"
  ax1.annotate(
      f'{t_p0:,} ({y_p0})'.replace(',', '.'),
      (x[i] - width / 2, t_p0),
      ha='center',
      va='bottom',
      fontsize=7.5,
      rotation=90,
      xytext=(0, 3),
      textcoords='offset points',
      color='#154360',
  )

  # Label Batang Oranye (previous > 0): Format "Total (Berhasil)"
  ax1.annotate(
      f'{t_pg0:,} ({y_pg0})'.replace(',', '.'),
      (x[i] + width / 2, t_pg0),
      ha='center',
      va='bottom',
      fontsize=7.5,
      rotation=90,
      xytext=(0, 3),
      textcoords='offset points',
      color='#b9770e',
  )

  # Label Garis Merah (Conversion Rate & Total Keberhasilan)
  success_tot = monthly['yes_total'].iloc[i]
  total_tot = monthly['total'].iloc[i]
  label_text = f'{success_tot}/{total_tot}\n({rate:.1f}%)'

  ax2.annotate(
      label_text,
      (x[i], rate),
      ha='center',
      va='bottom',
      fontsize=8,
      weight='bold',
      color='#c0392b',
      xytext=(0, 6),
      textcoords='offset points',
      bbox=dict(
          boxstyle='round,pad=0.25',
          facecolor='white',
          edgecolor='#e74c3c',
          alpha=0.9,
          linewidth=0.6,
      ),
  )

# Combine legends
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    loc='upper left',
    fontsize=11,
    frameon=True,
    facecolor='white',
    edgecolor='none',
    framealpha=0.9,
)

plt.title(
    'Performa Telemarketing dilihat dari nasabah yang pernah dihubungi vs belum pernah dihubungi (Mei 2008 - Nov 2010)',
    fontsize=15,
    weight='bold',
    pad=20,
)
plt.tight_layout()
# plt.savefig('previous_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

Hasil Visualisasi Diagram Batang

- Batang biru (previous = 0): menghitung nasabah baru (true cold leads) yang belum pernah dihubungi oleh kampanye manapun sebelumnya.

- Batang Oranye (previous > 0): Menampilkan seluruh nasabah yang memiliki rekam jejak pernah dihubungi pada masa sebelumnya (termasuk yang sebelumnya berstatus gagal/menolak.

Pada Mei–Agustus 2008, grafik menunjukkan hanya terlihat batang biru tanpa adanya batang oranye sama sekali, menandakan bahwa kampanye tahap awal tersebut 100% menargetkan nasabah baru (cold acquisition). Kemudian di bulan-bulan lainnya, Bank mengubah taktiknya dengan menghubungi nasabah yang pernah dihubungi sebelumnya (previous > 0), dan terlihat secara keseluruhan ada peningkatan conversion rate.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Pastikan df_bank sudah dimuat di workspace Anda
# df_bank = pd.read_csv('bank_cleaned.csv', sep=';')

# Pemetaan Bulan untuk Urutan Kronologis
month_map = {
    'mar': 3,
    'apr': 4,
    'may': 5,
    'jun': 6,
    'jul': 7,
    'aug': 8,
    'sep': 9,
    'oct': 10,
    'nov': 11,
    'dec': 12,
}
month_names_id = {
    3: 'Mar',
    4: 'Apr',
    5: 'Mei',
    6: 'Jun',
    7: 'Jul',
    8: 'Agu',
    9: 'Sep',
    10: 'Okt',
    11: 'Nov',
    12: 'Des',
}

# 2. Agregasi tanpa membuat kolom baru permanen di DataFrame utama
df_temp = df_bank.copy()
df_temp['month_num'] = df_temp['month'].map(month_map)

monthly_contact = (
    df_temp.groupby(['year', 'month_num'])
    .agg(
        total_cellular=('contact', lambda x: (x == 'cellular').sum()),
        total_telephone=('contact', lambda x: (x == 'telephone').sum()),
        yes_cellular=(
            'y',
            lambda x: (
                (df_temp.loc[x.index, 'contact'] == 'cellular') & (x == 'yes')
            ).sum(),
        ),
        yes_telephone=(
            'y',
            lambda x: (
                (df_temp.loc[x.index, 'contact'] == 'telephone') & (x == 'yes')
            ).sum(),
        ),
        total=('y', 'count'),
        yes_total=('y', lambda x: (x == 'yes').sum()),
    )
    .reset_index()
)

monthly_contact = monthly_contact.sort_values(
    ['year', 'month_num']
).reset_index(drop=True)
monthly_contact['period'] = (
    monthly_contact['month_num'].map(month_names_id)
    + ' '
    + monthly_contact['year'].astype(str)
)

# 3. Membuat Visualisasi Grouped Bar Chart
sns.set_theme(style='whitegrid')
fig, ax1 = plt.subplots(figsize=(18, 9))

x = np.arange(len(monthly_contact))
width = 0.38

# Plot Batang Cellular (Biru) dan Telephone (Ungu)
rects1 = ax1.bar(
    x - width / 2,
    monthly_contact['total_cellular'],
    width,
    label="contact = 'cellular' (Ponsel)",
    color='#2980b9',
    alpha=0.9,
)
rects2 = ax1.bar(
    x + width / 2,
    monthly_contact['total_telephone'],
    width,
    label="contact = 'telephone' (Telepon Rumah)",
    color='#8e44ad',
    alpha=0.9,
)

ax1.set_ylabel(
    'Jumlah Panggilan / Kontak', fontsize=12, weight='bold', color='#2c3e50'
)
ax1.set_xlabel('Periode (Bulan - Tahun)', fontsize=12, weight='bold', labelpad=10)
ax1.set_xticks(x)
ax1.set_xticklabels(
    monthly_contact['period'], rotation=45, ha='right', fontsize=10
)
ax1.legend(fontsize=11, loc='upper right', frameon=True)
ax1.grid(True, linestyle='--', alpha=0.4, axis='y')

# MEMPERLUAS BATAS ATAS Y AGAR DIAGRAM TIDAK TERLIHAT PENUH/PADAT
max_val = max(
    monthly_contact['total_cellular'].max(),
    monthly_contact['total_telephone'].max(),
)
ax1.set_ylim(0, max_val * 1.5)

# 4. Menambahkan Anotasi Label pada Batang
for i in range(len(monthly_contact)):
  t_cell = monthly_contact['total_cellular'].iloc[i]
  y_cell = monthly_contact['yes_cellular'].iloc[i]
  t_tel = monthly_contact['total_telephone'].iloc[i]
  y_tel = monthly_contact['yes_telephone'].iloc[i]

  # Anotasi Cellular
  if t_cell > 0:
    ax1.annotate(
        f'{t_cell:,} ({y_cell})'.replace(',', '.'),
        (x[i] - width / 2, t_cell),
        ha='center',
        va='bottom',
        fontsize=7.5,
        rotation=90,
        xytext=(0, 4),
        textcoords='offset points',
        color='#1b4f72',
    )

  # Anotasi Telephone
  if t_tel > 0:
    ax1.annotate(
        f'{t_tel:,} ({y_tel})'.replace(',', '.'),
        (x[i] + width / 2, t_tel),
        ha='center',
        va='bottom',
        fontsize=7.5,
        rotation=90,
        xytext=(0, 4),
        textcoords='offset points',
        color='#512e5f',
    )

plt.title(
    'Komposisi Volume Telemarketing: Cellular vs Telephone (Mei 2008 - Nov'
    ' 2010)',
    fontsize=15,
    weight='bold',
    pad=20,
)
plt.tight_layout()
# plt.savefig('contact_comparison_fixed.png', dpi=300, bbox_inches='tight')
plt.show()
```

Dari diagram di atas terlihat bahwa pada mei-juni 2008, bank menghubungi nasabah ke nomor telepon rumah mereka (batang ungu), dan mulai Juli 2008 secara berkala ada transisi bank menghubungi nasabah ke telepon seluler mereka (handphone) dibandingkan menghubungi nomor telepon rumah mereka, dan terlihat ada peningkatan conversion rate.

Dengan menggunakan saluran seluler, telemarketer dapat menjangkau nasabah secara langsung dan privat. Hal ini menjadi salah satu pendorong utama mengapa tingkat keberhasilan (conversion rate) pada periode selanjutnya melonjak drastis dibandingkan masa-masa awal kampanye.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


# Pemetaan Bulan untuk Urutan Kronologis
month_map = {
    'mar': 3,
    'apr': 4,
    'may': 5,
    'jun': 6,
    'jul': 7,
    'aug': 8,
    'sep': 9,
    'oct': 10,
    'nov': 11,
    'dec': 12,
}
month_names_id = {
    3: 'Mar',
    4: 'Apr',
    5: 'Mei',
    6: 'Jun',
    7: 'Jul',
    8: 'Agu',
    9: 'Sep',
    10: 'Okt',
    11: 'Nov',
    12: 'Des',
}

# Membuat salinan agar tidak memodifikasi DataFrame asli
df_temp = df_bank.copy()
df_temp['month_num'] = df_temp['month'].map(month_map)

# 2. Agregasi Rata-rata Euribor & Conversion Rate
monthly = (
    df_temp.groupby(['year', 'month_num'])
    .agg(
        avg_euribor=('euribor3m', 'mean'),
        total=('y', 'count'),
        success=('y', lambda x: (x == 'yes').sum()),
    )
    .reset_index()
)

monthly['conversion_rate'] = (monthly['success'] / monthly['total']) * 100
monthly = monthly.sort_values(['year', 'month_num']).reset_index(drop=True)
monthly['period'] = (
    monthly['month_num'].map(month_names_id)
    + ' '
    + monthly['year'].astype(str)
)

# 3. Chart Creation
sns.set_theme(style='whitegrid')
fig, ax1 = plt.subplots(figsize=(18, 9))
x = np.arange(len(monthly))

# -- Plot Suku Bunga Euribor (Garis Biru) --
color_euribor = '#2980b9'
ax1.plot(
    x,
    monthly['avg_euribor'],
    color=color_euribor,
    marker='s',
    linewidth=3,
    markersize=8,
    label='Rata-rata Euribor 3M (%)',
)
ax1.set_ylabel(
    'Suku Bunga Euribor 3M (%)',
    fontsize=12,
    weight='bold',
    color=color_euribor,
)
ax1.set_xlabel('Periode', fontsize=12, weight='bold', labelpad=10)
ax1.set_xticks(x)
ax1.set_xticklabels(monthly['period'], rotation=45, ha='right', fontsize=10)
ax1.set_ylim(0, 6.5)  # Beri margin atas agar label tidak terpotong
ax1.grid(axis='y', linestyle='--', alpha=0.5)

# -- Plot Conversion Rate (Garis Merah) --
ax2 = ax1.twinx()
color_cr = '#e74c3c'
ax2.plot(
    x,
    monthly['conversion_rate'],
    color=color_cr,
    marker='o',
    linewidth=3,
    markersize=8,
    label='Conversion Rate (%)',
)
ax2.set_ylabel(
    'Conversion Rate (%)', fontsize=12, weight='bold', color=color_cr
)
ax2.set_ylim(0, 115)  # Beri margin atas untuk kotak anotasi
ax2.grid(False)

# 4. Menambahkan Anotasi
for i in range(len(monthly)):
  euri = monthly['avg_euribor'].iloc[i]
  cr = monthly['conversion_rate'].iloc[i]
  succ = monthly['success'].iloc[i]
  tot = monthly['total'].iloc[i]

  # Anotasi untuk Garis Biru (Euribor)
  ax1.annotate(
      f'{euri:.2f}%',
      (x[i], euri),
      ha='center',
      va='bottom',
      fontsize=8.5,
      weight='bold',
      color='#154360',
      xytext=(0, 7),
      textcoords='offset points',
  )

  # Anotasi untuk Garis Merah (Conversion Rate)
  label_text_cr = f'{succ:,}/{tot:,}\n({cr:.1f}%)'.replace(',', '.')
  ax2.annotate(
      label_text_cr,
      (x[i], cr),
      ha='center',
      va='bottom',
      fontsize=8.5,
      weight='bold',
      color='#922b21',
      xytext=(0, 9),
      textcoords='offset points',
      bbox=dict(
          boxstyle='round,pad=0.25',
          facecolor='white',
          edgecolor=color_cr,
          alpha=0.85,
          linewidth=0.6,
      ),
  )

# 5. Menggabungkan Legend
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(
    lines_1 + lines_2,
    labels_1 + labels_2,
    loc='upper center',
    fontsize=11,
    frameon=True,
    ncol=2,
)

plt.title(
    'Tren Suku Bunga Euribor 3M vs Conversion Rate (Mei 2008 - Nov 2010)',
    fontsize=15,
    weight='bold',
    pad=20,
)
plt.tight_layout()
# plt.savefig('euribor_trend_cr_fixed.png', dpi=300, bbox_inches='tight')
plt.show()
```

nilai `euribor3m` akan berpengaruh terhadap suku bunga deposito, semakin tinggi `euribor3m` maka semakin tinggi pula suku bunga deposito yang diberikan oleh bank. Maka asumsi umum adalah semakin tinggi euribor berarti semakin tinggi pula nasabah yang tertarik untuk membuka deposito berjangka, namun pada dataset justru kebalikannya, dan temuan ini sesuai dengan korelasi -0,31 pada diagram matriks korelasi sebelumnya.

- Pada awal krisis 2008, suku Bunga `Euribor3m` berada di puncak tertinggi (hampir menyentuh angka 5%). Akibatnya, Conversion Rate tertahan datar di kisaran 3% - 6%

- Pada transisi Desember 2008 - Maret 2009, suku bunga euribor mulai diturunkan secara tajam dan bertahap dalam merespon krisis ekonomi. Di saat yang sama conversion rate pun mengalami kenaikan.

- Memasuki pertengahan akhir 2009 hingga 2010, suku bunga `euribor3m` dipertahankan sangat rendah (stabil di kisaran 0,6% - 1,4%). Kondisi makro yang melunak ini direspons sangat positif oleh pasar, di mana Conversion Rate konsisten bertahan tinggi (antara 35% hingga menembus 50%).

Euribor 3M adalah salah satu indikator makro paling krusial untuk dipantau. Kampanye penawaran produk simpanan/deposito berjangka terbukti jauh lebih efektif dijalankan secara agresif pada masa ketika suku bunga antar bank sedang mengalami tren penurunan atau berada di level rendah (di bawah 1,5%).

##**Bagaimana menentukan strategi penawaran yang efektif agar bank tidak perlu menghubungi nasabah secara acak?**


Berdasarkan seluruh identifikasi dan analisis yang telah dilakukan terhadap data demografi nasabah, historis telemarketing dan indikator sosial-ekonomi, maka bisa diperoleh bagaimana menyusun strategi telemarketing yang efektif dengan memprioritaskan nasabah yang tinggi hasil konversinya

Nasabah yang setuju untuk mengambil deposito pada kampanye telemarketing sebelumnya `poutcome` = success, ditawarkan kembali pada kampanye berikutnya, karena berdasarkan hasil analisis, mereka memiliki kecendrungan tinggi (~65%) untuk berhasil dikonversi kembali (setuju untuk membuka deposito berjangka lagi).

Kemudian prioritaskan nasabah yang pernah dihubungi pada masa sebelumnya untuk dihubungi kembali, kelompok ini memiliki kecenderungan lebih tinggi untuk dikonversi dibandingkan nasabah yang belum pernah dihubungi sama sekali. Namun bukan berarti melupakan nasabah baru/nasabah yang tidak pernah dihubungi sama sekali, karena mereka bisa memiliki potensi lebih besar untuk dikonversi di kampanye yang berikutnya lagi.

Telemarketer harus menghubungi nasabah ke nomor ponsel mereka, hindari menelepon ke telepon rumah mereka jika memungkinkan. `contact` = cellular

Temuan berdasarkan temuan demografi/karakteristik nasabah

- Berdasarkan usia, prioritaskan kelompok low volume - high conversion (`>60` dan `<25`), kemudian kelompok usia `56-60`, lalu kelompok high volume - low conversion yang didominasi oleh usia produktif (`25-35` dan `36-55`) karena mayoritas dataset didominasi kelompok ini.

- Berdasarkan pekerjaan, prioritaskan kelompok retired dan student (karena masih terkait usia sesuai temuan sebelumnya), lalu prioritaskan pada kelompok pekerjaan yang memiliki conversion rate tertinggi hingga terendah.

- Fokuskan kepada nasabah yang memiliki tingkat pendidikan terlebih dahulu, karena berdasarkan temuan analisis, semakin tinggi pendidikan akhir nasabah, cenderung untuk setuju membuka deposito berjangka.

Faktor lain yang perlu dipertimbangkan karena dapat mempengaruhi keberhasilan konversi
- Secara umum batasi frekuensi panggilan `campaign range` ke nasabah maksimal 3 kali karena lebih dari itu bisa membuat nasabah menjadi jenuh, berdasarkan hasil temuan terlihat bahwa semakin banyak melakukan panggilan ke nasabah cenderung semakin rendah conversion ratenya, patokan ini juga serupa berdasarkan statistik deskriptif yang didapatkan Q1=1 kali, Q2 = 2 kali dan Q3= 3 kali. Namun pembatasan ini juga jangan diikuti secara kaku/rigid, karena bisa saja pada kasus tertentu telemarketer membutuhkan lebih dari 3 kali untuk menelepon nasabah.

- Dari temuan analisis terhadap tren `euribor3m` didapatkan bahwa penurunan suku bunga euribor3m ini justru berdampak terhadap peningkatan conversion rate. Maka pada saat suku bunga euribor3m sedang rendah terutama berkisar 0,6%-1,4% Bank bisa memanfaatkan peluang ini sebagai kesempatan emas oleh tim telemarketing secara gencar untuk memperoleh nasabah membuka deposito berjangka. Dari segi bisnis pun hal ini selaras dengan profit bank (lihat bagian pendahuluan terutama sepanjang tahun 2000-2007), karena rendahnya suku bunga juga bisa berpengaruh terhadap peningkatan demand/pengajuan pinjaman ke bank.

# **SIMULASI CONVERSION RATE BERDASARKAN REKOMENDASI**

Berdasarkan rekomendasi sebelumnya (rekomendasi kedua), selanjutnya akan dilakukan perhitungan secara kasar untuk mensimulasikan hasil conversion rate apabila menggunakan seleksi berdasarkan temuan/rangkaian hasil analisis namun penyeleksian dilakukan tidak secara rigid/kaku.

Berdasarkan grafik **Performa Telemarketing dan conversion rate sepanjang 2008-2010** sebelumnya, akan terlihat batang biru yang mengindikasikan gagal terkonversi `y` = no, di sini akan diambil mulai dari Mei 2008 hingga Mei 2009.

```python

# 1. Pemetaan urutan bulan kalender ke angka (sebagai variabel sementara)
month_map = {
    'mar': 3, 'apr': 4, 'may': 5, 'jun': 6, 'jul': 7, 'aug': 8,
    'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12
}

month_num = df_bank['month'].str.lower().map(month_map)

# 2. Kondisi filter periode (Mei 2008 s.d. Mei 2009) dan target y == 'no'
cond_period = (
    ((df_bank['year'] == 2008) & (month_num >= 5)) |
    ((df_bank['year'] == 2009) & (month_num <= 5))
)

# 3. Menyimpan hasil filter ke df_potential
df_potential = df_bank[cond_period & (df_bank['y'] == 'no')].copy()

# 4. Verifikasi hasil
print(f"Total baris df_potential: {len(df_potential):,}")
print("\nDistribusi per tahun:")
print(df_potential['year'].value_counts())
print("\nVerifikasi nilai kolom 'y':")
print(df_potential['y'].value_counts())
```

Selanjutnya hanya akan diambil pada nasabah yang belum pernah dihubungi saja.

```python
# Menyaring hanya nasabah yang belum pernah dihubungi sebelumnya (previous == 0)
df_potential = df_potential[df_potential['previous'] == 0].copy()

# Verifikasi hasil
print(f"Jumlah baris df_potential saat ini: {len(df_potential):,} baris")
print("\nDistribusi nilai kolom previous:")
print(df_potential['previous'].value_counts())
```

Walaupun nasabah ini pada 2008-2010 tidak pernah dihubungi sebelumnya, namun pada campaign marketing berikutnya secara otomatis nasabah ini akan tercatat sebagai nasabah yang pernah dihubungi pada masa berikutnya.

Lalu, sebagian besar nasabah ini dihubungi ke nomor telepon rumahnya, maka Bank harus menghubungi ke nomor `celllular` mereka. Diasumsikan catatan Bank memiliki beberapa nomor yang bisa dihubungi, terutama nomor `cellular`.

Berikutnya kita akan menyeleksi lagi nasabah berdasarkan jenis pekerjaannya, akan dihapus lagi mereka dengan pekerjaan `blue-collar`, `housemaid`, `unemployed` dengan asumsi jenis pekerjaan ini memiliki kecendrungan lebih susah untuk dikonversi dan asumsi minim penghasilannya. Walaupun `unemployed` memiliki kecendrungan konversi yang tinggi, namun di sini kita akan drop saja dengan asumsi unemployed artinya tidak ada penghasilan (walaupun pada kenyataannya belum tentu seperti itu, bisa saja orang kaya atau lagi sedang mencari pekerjaan lagi, atau lansia).

```python
# 1. Daftar pekerjaan yang ingin dihapus
excluded_jobs = ['blue-collar', 'housemaid', 'unemployed']

# 2. Filter df_potential untuk mengecualikan pekerjaan tersebut
df_potential = df_potential[~df_potential['job'].isin(excluded_jobs)].copy()

# 3. Verifikasi hasil pemfilteran
print(f"Sisa baris di df_potential: {len(df_potential):,}")
print("\nDistribusi pekerjaan yang tersisa:")
print(df_potential['job'].value_counts())
```

Selanjutnya kita akan seleksi mereka yang hanya memiliki 3 tingkat pendidikan teratas. Dengan asumsi semakin tinggi pendidikan semakin tinggi kecendrungan conversion rate-nya sesuai hasil temuan analisis sebelumnya.

Sedangkan jika diambil nilai conversion rate (berdasarkan total) sepanjang juni 2009 hingga akhir

```python
# 1. Pemetaan urutan dan nama bulan
month_map = {
    'mar': 3, 'apr': 4, 'may': 5, 'jun': 6, 'jul': 7, 'aug': 8,
    'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12
}
month_names_id = {
    3: 'Mar', 4: 'Apr', 5: 'Mei', 6: 'Jun', 7: 'Jul', 8: 'Agu',
    9: 'Sep', 10: 'Okt', 11: 'Nov', 12: 'Des'
}

# 2. Filter periode Juni 2009 hingga November 2010 tanpa mengubah df_bank
month_num = df_bank['month'].str.lower().map(month_map)
cond = (
    ((df_bank['year'] == 2009) & (month_num >= 6)) |
    ((df_bank['year'] == 2010) & (month_num <= 11))
)

subset = df_bank[cond].copy()
subset['month_num'] = month_num[cond]

# 3. Agregasi per bulan
rincian = subset.groupby(['year', 'month_num']).agg(
    total_panggilan=('y', 'count'),
    berhasil_yes=('y', lambda x: (x == 'yes').sum())
).reset_index()

rincian = rincian.sort_values(['year', 'month_num']).reset_index(drop=True)
rincian['conversion_rate'] = (rincian['berhasil_yes'] / rincian['total_panggilan']) * 100
rincian['periode'] = rincian['month_num'].map(month_names_id) + ' ' + rincian['year'].astype(str)

# 4. Format tabel tampilan
tabel_rincian = rincian[['periode', 'total_panggilan', 'berhasil_yes', 'conversion_rate']].copy()
tabel_rincian.columns = ['Periode', 'Total Panggilan', 'Berhasil (yes)', 'Conversion Rate (%)']
tabel_rincian['Conversion Rate (%)'] = tabel_rincian['Conversion Rate (%)'].map(lambda x: f"{x:.2f}%")

# 5. Tambahkan baris total agregat keseluruhan periode
total_calls = tabel_rincian['Total Panggilan'].sum()
total_yes = tabel_rincian['Berhasil (yes)'].sum()
total_rate = (total_yes / total_calls) * 100

baris_total = pd.DataFrame([{
    'Periode': 'Total Periode',
    'Total Panggilan': total_calls,
    'Berhasil (yes)': total_yes,
    'Conversion Rate (%)': f"{total_rate:.2f}%"
}])

tabel_akhir = pd.concat([tabel_rincian, baris_total], ignore_index=True)
print(tabel_akhir.to_string(index=False))
```
